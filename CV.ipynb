```python
import yaml
from pathlib import Path

src = Path("/content/EMRDM/configs/example_training/cuhk.yaml")
dst = Path("/content/EMRDM/configs/example_training/cuhk_colab_baseline.yaml")

with open(src, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

base = "/content/EMRDM/data/C-CUHK/C-CUHK - 副本"

# Released checkpoint
config["model"]["params"]["ckpt_path"] = "/content/EMRDM/checkpoints/last.ckpt"

# Dataset paths
for split in ["train", "validation", "test", "predict"]:
    config["data"]["params"][split]["params"]["datasets_dir"] = f"{base}/CUHK-CR1"
    config["data"]["params"][split]["params"]["nir_datasets_dir"] = f"{base}/nir/CUHK-CR1"

# Match released test sampler settings
sampler = config["model"]["params"]["sampler_config"]["params"]
sampler["num_steps"] = 4
sampler["s_churn"] = 0.1
sampler["s_tmin"] = 0.0
sampler["s_tmax"] = 100000000.0
sampler["s_noise"] = 0.995

# One GPU: your Tesla T4
config["lightning"]["trainer"]["devices"] = 1

with open(dst, "w", encoding="utf-8") as f:
    yaml.safe_dump(config, f, sort_keys=False)

print("Created:")
print(dst)
```
